```python
import parasail

# BLOSUM62 matrix from parasail (already integer-scaled)
PARASAIL_MATRIX = parasail.blosum62

# gap_open and gap_extend are positive penalties in parasail
# pairwise2: -10 (open), -0.5 (extend)
# we’ll approximate with 10 / 1, or you can adjust if needed.
GAP_OPEN = 10
GAP_EXTEND = 1

def align_and_identity_longest_parasail(seq1: str, seq2: str) -> float:
    """
    Global alignment with parasail (Needleman-Wunsch) + identity over longest original length.
    """
    if not seq1 and not seq2:
        return 0.0

    # nw_trace_* gives us traceback strings, needed for identity
    # _scan_16 is a good default; you can also try _striped_16
    result = parasail.nw_trace_scan_16(
        seq1,
        seq2,
        GAP_OPEN,
        GAP_EXTEND,
        PARASAIL_MATRIX,
    )

    aligned1 = result.traceback.query  # aligned query (with '-')
    aligned2 = result.traceback.ref    # aligned ref (with '-')

    matches = 0
    for a, b in zip(aligned1, aligned2):
        if a != "-" and b != "-" and a == b:
            matches += 1

    denom = max(len(seq1), len(seq2))
    return matches / denom if denom else 0.0

def worker_batch(batch):
    out = []
    for seq, train_seq in batch:
        out.append(align_and_identity_longest_parasail(seq, train_seq))
    return out

```
